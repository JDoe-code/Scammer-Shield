```python
import pandas as pd
```

```python
PhishingEmailV2Data = pd.read_csv(
    "phishing_emailDetectionv2.csv",engine="python",on_bad_lines="skip"
)
```

```python

# Select the five important columns
important_columns = [
     "url_domain",
    "contains_link",
    "email_body",
    "subject",
    "sender_domain",
    "sender_email",
    "spoofed_domain",
    "label"

]

# Create a separate DataFrame
df_clean = PhishingEmailV2Data[important_columns].copy()

# Display the first 10 rows
display(df_clean.head(10))

# Check rows and columns
print("Shape:", df_clean.shape)

print(df_clean.columns.tolist())
```

```python
PhishingEmailV2Data.duplicated().sum()

df_clean.duplicated(
    subset=["subject", "email_body"]
).sum()

```

```python

# Create a DataFrame containing only phishing emails
phishing_df = df_clean[
    df_clean["label"] == "Phishing"
].copy()

# Create a DataFrame containing only legitimate emails
legit_df = df_clean[
    df_clean["label"] == "Legitimate"
].copy()

# Reset indexes
phishing_df = phishing_df.reset_index(drop=True)
legit_df = legit_df.reset_index(drop=True)

# Display the number of emails
print("Phishing emails:", phishing_df.shape)
print("Legitimate emails:", legit_df.shape)

# Verify all rows were classified
print("Original rows:", len(df_clean))
print("Split rows:", len(phishing_df) + len(legit_df))
```

```python

for col in ["contains_link", "spoofed_domain"]:
    df_clean[col] = df_clean[col].replace({
        "Yes": 1,
        "No": 0
    })

# Display updated DataFrame
display(df_clean.head())
```

```python
#verify the spoof_domain and Contains_link
print(df_clean[["contains_link", "spoofed_domain"]].head(10))
```

```python
# Reset Index
df_clean = df_clean.reset_index(drop =True)
```

```python
#Diplay The cleaned DataFrame
display(df_clean.head())
print("Shape:", df_clean.shape)
```

```python

from pathlib import Path

# Create the output folder if it does not exist
output_folder = Path("../done")
output_folder.mkdir(parents=True, exist_ok=True)

# Split the cleaned DataFrame into two groups
legit_df = df_clean[
    df_clean["label"] == "Legitimate"
].copy()

phishing_df = df_clean[
    df_clean["label"] == "Phishing"
].copy()

# Reset row indexes
legit_df = legit_df.reset_index(drop=True)
phishing_df = phishing_df.reset_index(drop=True)

# Verify that every row belongs to one of the two groups
assert len(legit_df) + len(phishing_df) == len(df_clean)

# Create the two CSV files
legit_df.to_csv(
    output_folder / "clean_phishing_email2.csv",
    index=False
)

phishing_df.to_csv(
    output_folder / "real_phishing_email2.csv",
    index=False
)

print("Both CSV files created successfully!")
print("Legitimate emails:", len(legit_df))
print("Phishing emails:", len(phishing_df))
```
