```python
import pandas as pd
```

```python
legit_df = pd.read_csv("legit.csv")
#print(legit_df.head())
# print(legit_df.columns)
# legit_df.isna().sum()
# import os
# print(os.getcwd())
# print(os.listdir("."))
```

```python
cleaned_legit_df = pd.DataFrame(columns=['body','subject','urls','label'])
print(cleaned_legit_df.columns)
cleaned_legit_df = legit_df[['body', 'subject', 'urls', 'label']].copy()
print(cleaned_legit_df.head())

```

```python
phishing_df = pd.read_csv('phishing.csv')
print(phishing_df.columns)
cleaned_phishing_df = pd.DataFrame(columns=['body','subject','urls','label'])
print(cleaned_phishing_df.columns)
cleaned_phishing_df = phishing_df[['body', 'subject', 'urls', 'label']].copy()
print(cleaned_phishing_df.head())
```

```python
ceas_df = pd.read_csv('CEAS_08.csv')
# ceas_df.head()
# ceas_df.isna().sum()
print(ceas_df.columns)
cleaned_ceas_df = pd.DataFrame(columns=['body','subject','urls','label'])
print(cleaned_ceas_df.columns)
cleaned_ceas_df = ceas_df[['body', 'subject', 'urls', 'label']].copy()
print(cleaned_ceas_df.head())
```

```python
ceas_legit = cleaned_ceas_df[cleaned_ceas_df['label'] == 0]
ceas_phishing = cleaned_ceas_df[cleaned_ceas_df['label'] == 1]
# ceas_phishing.head()
# ceas_legit.head()

cols = ['body', 'subject', 'urls', 'label']
cleaned_legit_df['source'] = 'legit_csv'
cleaned_phishing_df['source'] = 'phishing_csv'
ceas_legit['source'] = 'ceas_08'
ceas_phishing['source'] = 'ceas_08'

final_legit = pd.concat([cleaned_legit_df, ceas_legit[cols + ['source']]], ignore_index=True)
final_phishing = pd.concat([cleaned_phishing_df, ceas_phishing[cols + ['source']]], ignore_index=True)

final_legit = final_legit.dropna(subset=['body']).drop_duplicates(subset=['body', 'subject'])
final_phishing = final_phishing.dropna(subset=['body']).drop_duplicates(subset=['body', 'subject'])

print(len(final_legit), len(final_phishing))
print(final_legit['label'].unique(), final_phishing['label'].unique())

final_legit.to_csv("../done/legit_cleaned.csv", index=False)
final_phishing.to_csv("../done/phishing_cleaned.csv", index=False)


```
